# Importation des données opérationnelles

```python
xlsx = pd.ExcelFile("C:/Users/testa/Documents/Zalya_Stage/DATA/REEL DATA.xlsx")
print(xlsx.sheet_names)
```

```python
df_operations = pd.read_excel(
 "C:/Users/testa/Documents/Zalya_Stage/DATA/REEL DATA.xlsx",
    sheet_name="operations "
)
```

```python
df_operations.head()
```

```python
# Standardisation des colonnes
df_operations.columns = (
    df_operations.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("/", "_")
)
```

```python
# Rénnomation des colonnes
df_operations.rename(columns={
    "date_de_facture": "date_facture",
    "bénéficiaire": "beneficiaire",
}, inplace=True)
```

```python
# Supprésion de la derniere ligne total 
df_operations = df_operations[
    df_operations["service"].notna()
]
```

```python
df_operations
```

## Nettoyage des montants

```python
# colonnes financière
colonnes_montants = [
    "montant_ht",
    "tva",
    "ca",
    "montant_ttc"
]
```

```python
# nettoyage
for col in colonnes_montants:
    df_operations[col] = (
        df_operations[col]
        .astype(str)
        .str.replace("\xa0", "", regex=False)
        .str.replace(" ", "", regex=False)
        .str.replace(",", ".", regex=False)
    )
    df_operations[col] = pd.to_numeric(
        df_operations[col],
        errors="coerce"
    )
```

## Conversion  des dates

```python
df_operations["date_facture"] = pd.to_datetime(
    df_operations["date_facture"],
    dayfirst=True,
    errors="coerce"
)
```

```python
df_operations["annee"] = (
    df_operations["date_facture"]
    .dt.year
)
df_operations["mois"] = (
    df_operations["date_facture"]
    .dt.month
)
df_operations["trimestre"] = (
    df_operations["date_facture"]
    .dt.quarter
)
```

## Uniformisation du texe

```python
colonnes_textes = [
    "service",
    "description",
    "beneficiaire"
]
```

```python
# uniformisation
for col in colonnes_textes:
    df_operations[col] = (
        df_operations[col]
        .astype(str)
        .str.strip()
        .str.lower()
    )
```

```python
df_operations
```

## Création d'une catégorie analytique financière

```python
categories_depenses = {

    "fourniture bureau":
    "administratif",

    "autres achats":
    "divers",

    "travaux":
    "maintenance",

    "travaux maintenance":
    "maintenance",

    "prestataire":
    "prestations",

    "transport":
    "divers",

    "internet":
    "internet",

    "loyer":
    "logement",

    "communication":
    "administratif",

    "charges personnel":
    "divers",

    "electricité":
    "electricite"

}
```

```python
df_operations["categorie_depense"] = (
    df_operations["service"]
    .map(categories_depenses)
)
```

```python
df_operations
```

```python
# montant_nul
df_operations[
    df_operations["montant_ttc"] <= 0
]
```

```python
# fournisseur manquant
df_operations[
    df_operations["beneficiaire"].isnull()
]
```

```python
# montant négatif
df_operations[
    df_operations["montant_ttc"] < 0
]
```

```python
# ID operation
df_operations["operation_id"] = range(
    1,
    len(df_operations) + 1
)
```
