## -Valeurs manquantes

```python
df_ventes["Date paiement"]= df_ventes["Date paiement"].fillna("Vide")
df_ventes["Mode paiement"]= df_ventes["Mode paiement"].fillna("Vide")
df_ventes["N° Facture"]= df_ventes["N° Facture"].fillna("Vide")
```

## - Nettoyae des montants

```python
df_ventes["Montant HT"].describe
```

```python
df_ventes["Montant HT"].dtype
```

```python
df_ventes["Montant HT"] = df_ventes["Montant HT"].replace(["", "NA", "-", "nan"], None)
df_ventes["Montant HT"] = pd.to_numeric(df_ventes["Montant HT"], errors="coerce")
```

```python
df_ventes["Montant HT"].describe()
```

## -Traitement des dates 

```python
#Conversion
df_ventes["Date paiement"] = pd.to_datetime(
    df_ventes["Date paiement"],
    errors="coerce"
)
```

```python
df_ventes["annee"] = (
    df_ventes["Date paiement"]
    .dt.year
)

df_ventes["mois"] = (
    df_ventes["Date paiement"]
    .dt.month
)
df_ventes["jour"] = (
    df_ventes["Date paiement"]
    .dt.day
)
```

## -Uniformisation du texte

```python
df_ventes["Client (s)"] = (
    df_ventes["Client (s)"]
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
    .str.replace(r"[^a-zA-Z0-9\s]", "", regex=True)
)

df_ventes["Service"] = (
    df_ventes["Service"]
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
    .str.replace(r"[^a-zA-Z0-9\s]", "", regex=True)
)
```

```python
df_ventes.head()
```

## -Revu des colonnes

```python
df_ventes.columns = df_ventes.columns.str.strip()
df_ventes.rename(columns={
    "Date de Facture": "Date_facture",
    "Date d'écheance": "Date_echeance",
    "N° Facture": "Num_facture",
    "Client (s)": "Clients",
    "Montant HT": "Montant_HT",
    "Montant TTC" : "Montant_TTC",
    "Date paiement" : "Date_paiement",
    "Mode paiement" : "Mode_paiement"
}, inplace=True)
```
