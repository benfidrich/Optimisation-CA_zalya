### Uniformisation du texte 

```python
df_clients["Clients"] = (
    df_clients["Clients"]
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
    .str.replace(r"[^a-zA-Z0-9\s]", "", regex=True)
)
```

```python
df_clients.rename(columns={
    " Téléphone": "Telephone",
    "E-mail": "E_mail"
}, inplace=True)
```

```python
df_clients["E_mail"]= df_clients["E_mail"].fillna("Vide")
df_clients["Telephone"]= df_clients["Telephone"].fillna("Vide")
```

```python
df_clients.head()
```

### Vérification de clients sans nom

```python
df_clients[ df_clients["Clients"].isnull()]
```

```python
# Création d'une clé pour les liens avec les autres entités
df_clients["client_id"] = range(  1,len(df_clients) + 1)
```

```python
df_clients
```

```python
df_clients.rename(columns={
    "Clients": "client",
    "E_mail": "email",
    "Telephone": "telephone",
    "Adresse ": "adresse"
}, inplace=True)
```

```python
colonnes_texte = [
    "client",
    "adresse"
]
for col in colonnes_texte:
    if col in df_clients.columns:
        df_clients[col] = (
            df_clients[col]
            .astype(str)
            .str.strip()
            .str.lower()
        )
```
