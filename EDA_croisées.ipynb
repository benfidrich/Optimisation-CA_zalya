## Analyse financière mensuelle

```python
# Revenu mensuel
revenus_mensuels = (
    df_ventes
    .groupby(["annee", "mois"])["montant_ttc"]
    .sum()
)
```

```python
# Dépense mensuelle
depenses_mensuelles = (
    df_operations
    .groupby(["annee", "mois"])["montant_ttc"]
    .sum()
)
```

```python
finance = pd.concat(
    [revenus_mensuels, depenses_mensuelles],
    axis=1
)
finance.columns = [
    "revenus",
    "depenses"
]
print(finance)
```

```python
# Profit
finance["profit"] = (
    finance["revenus"]
    - finance["depenses"]
)
finance["profit"]
```

```python
# Ration rentabilité 
finance["ratio_rentabilite"] = (
    finance["profit"]
    / finance["revenus"]
) * 100
print(finance["ratio_rentabilite"])
```

***Profit négatif (danger), forte hausse dépense (anomalie)***

## Analyse des dépenses critiques

```python
# Pourcentage dépenses
pourcentage_depenses = (
    depenses_categories
    / depenses_categories.sum()
) * 100
print(pourcentage_depenses)
```
