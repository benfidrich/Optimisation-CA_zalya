```python
import pandas as pd
```

# Importation des données de devis et commandes

```python
df_devis_commandes = pd.read_excel("C:/Users/testa/Documents/Zalya_Stage/DATA/REEL DATA.xlsx", sheet_name = "Dévis et bon de commande")
```

```python
df_devis_commandes.head()
```

```python
df_devis_commandes.columns = (
    df_devis_commandes.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("/", "_")
)
```

```python
print(df_devis_commandes.columns)
```

```python
df_devis_commandes.rename(columns={
    "référence_commande": "reference_commande",
    "date_de_création": "date_creation",
    "total": "montant",
    "activités": "activite"
}, inplace=True)
```

```python
print(df_devis_commandes.columns)
```

```python
# Nettoyage des montants
df_devis_commandes["montant"] = (
    df_devis_commandes["montant"]
    .astype(str)
    .str.replace("\xa0", "", regex=False)
    .str.replace(" ", "", regex=False)
    .str.replace(",", ".", regex=False)
)
```

```python
df_devis_commandes["montant"] = pd.to_numeric(
    df_devis_commandes["montant"],
    errors="coerce"
)
```

```python
df_devis_commandes.head()
```

```python
# Conversion des dates
df_devis_commandes["date_creation"] = pd.to_datetime(
    df_devis_commandes["date_creation"],
    dayfirst=True,
    errors="coerce"
)
```

```python
df_devis_commandes["annee"] = (
    df_devis_commandes["date_creation"]
    .dt.year
)
df_devis_commandes["mois"] = (
    df_devis_commandes["date_creation"]
    .dt.month
)
df_devis_commandes["trimestre"] = (
    df_devis_commandes["date_creation"]
    .dt.quarter
)
```

### Uniformisation du texte 

```python
df_devis_commandes["client"] = (
    df_devis_commandes["client"]
    .astype(str)
    .str.strip()
    .str.upper()
)
```

```python
df_devis_commandes["vendeur"] = (
    df_devis_commandes["vendeur"]
    .astype(str)
    .str.strip()
    .str.upper()
)
```

```python
# Statut 
df_devis_commandes["statut"] = (
    df_devis_commandes["statut"]
    .astype(str)
    .str.strip()
    .str.lower()
)
```

```python
df_devis_commandes.head()
```

```python
# variable analytique
df_devis_commandes["type_transaction"] = (
    df_devis_commandes["statut"]
)
```

```python
# Indicateur de convertion 
df_devis_commandes["conversion"] = (
    df_devis_commandes["statut"]
    == "commande client"
)
```

```python
df_devis_commandes.head()
```

```python
#Suppression de la colonne activite
# df_devis_commandes= df_devis_commandes.drop("activite", axis= 1)
```

```python
# référence dupliquée
df_devis_commandes[
    df_devis_commandes["reference_commande"]
    .duplicated()
]
```

```python
# ID
df_devis_commandes["transaction_id"] = range(
    1,
    len(df_devis_commandes) + 1
)
```

```python
colonnes_texte = [
    "client",
    "vendeur"
]
for col in colonnes_texte:
    if col in df_devis_commandes.columns:
        df_devis_commandes[col] = (
            df_devis_commandes[col]
            .astype(str)
            .str.strip()
            .str.lower()
        )
```
