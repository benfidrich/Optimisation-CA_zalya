### Prétraitement des données

```python
df_p_s.rename(
    columns={
        "Produits /Services":
        "nom_service"
    },
    inplace=True
)
```

```python
df_p_s["service_id"] = range(
    1,
    len(df_p_s) + 1
)
```

```python
df_p_s.head()
```

#### Ajout d'une colonne catégorie

```python
categories = {

    "conception web": "digital",

    "développement plateformes ou logiciels":
    "digital",

    "communication digitale":
    "marketing",

    "imprimerie":
    "print_media",

    "imporation":
    "logistique",

    "gestion et vente des biens":
    "commerce",

    "prestations industrielles ( climatisation et froid, maintenance industrielle, peinture,l'électricité)":
    "industrie",

    "mise à disposition du personnel":
    "rh",

    "gestion de sécurité":
    "securite",

    "formation présentiel et en ligne":
    "formation"
}
```

```python
df_p_s["categorie"] = (
    df_p_s["nom_service"]
    .map(categories)
)
```

```python
df_p_s.head()
```

```python
df_p_s["actif"] = True
```

```python
df_p_s["prix_reference"] = None
df_p_s["cout_estime"] = None
df_p_s["marge_estimee"] = None
```

```python
df_p_s.head()
```

```python
df_p_s= df_p_s.drop("N°",axis= 1)
```

```python
df_p_s.head()
```

```python
df_p_s["nom_service"]= df_p_s["nom_service"].replace({
    "communication digitale": "communication",
    "imporation": "importation",
})
```

```python
df_p_s
```

```python
df_p_s["prix_reference"] = (
    df_p_s["prix_reference"]
    .fillna(0)
)
df_p_s["cout_estime"] = (
    df_p_s["cout_estime"]
    .fillna(0)
)
df_p_s["marge_estimee"] = (
    df_p_s["marge_estimee"]
    .fillna(0)
)
```
