# III - KPI clés

**Nombre total d'opportunités*

```python
total_opportunites = len(df)

print(total_opportunites)
```

**Opportunité à forte probablité*

```python
fortes_prob = df[df["Probabilite"] >= 70]
nombre_fortes = len(fortes_prob)
print(nombre_fortes)
```

**Revenu attendu total*

```python
revenu_total = df["Revenu_attendu"].sum()
print(revenu_total)
```

**Valeur moyenne des opportunités*

```python
valeur_moyenne_op = df["Revenu_attendu"].mean()
print(valeur_moyenne_op)
```

**Opportunité par étapes*

```python
kpi_etapes = df["Etape"].value_counts()
print(kpi_etapes)
```

**Revenu par étape*

```python
revenu_etape = df.groupby("Etape")["Revenu_attendu"].sum()
print(revenu_etape)
```

**Performance par vendeur*

```python
perf_vendeur = df.groupby("Vendeur")["Revenu attendu"].sum()
print(perf_vendeur)
```

**Performance par client*

```python
clients = df.groupby("Client")["Revenu attendu"].sum()
print(clients.sort_values(ascending=False))
```

**Opportunités critiques/non critiques/moyennes*

```python
critiques = df[df["Priorite"] == "Haute"]
print(len(critiques))
non_critique = df[df["Priorite"] =="Basse"]
print(len(non_critique))
moyenne = df[df["Priorite"] =="Moyenne"]
print(len(moyenne))
```

**Taux d'opportunité gagnée/perdue*

```python
taux_gp = df["Gagnée/Perdue"].value_counts(normalize=True) * 100
print(taux_gp)
```

```python
op_gagnees = len(df[df["Gagnee_Perdue"] == "Gagné"])
taux_conversion = (op_gagnees / total_opportunites) * 100
print("Taux de conversion :", taux_conversion)
```
